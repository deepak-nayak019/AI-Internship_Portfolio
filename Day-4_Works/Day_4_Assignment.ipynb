1.Full Pipeline: Clean the Orders Dataset

```python
import pandas as pd
import numpy as np
#load data
df = pd.read_csv("data.csv")
df
```

```python
#Convert fake null value to naN
df = df.replace(r'^\s*$', np.nan,regex=True)
```

```python
df.info()
```

```python
print(df.columns.tolist())
```

```python
#*Remove Extra space from column*(Main reason this entire data):

df.columns = df.columns.str.strip()

print(df.columns.tolist)
```

```python
#Clean text columns:

text_column = ['customer','city','status']
for col in text_column:
    df[col] = df[col].str.strip()
```

```python
#Standardize city name:

df['city'] = df['city'].str.title()
```

```python
df
```

```python
#Convert amount to numeric:

df.dtypes
```

```python
df['amount'] = pd.to_numeric(df['amount'],errors='coerce')
```

```python
df.dtypes
```

```python
df
```

```python
#Chck missing value and duplicate:
print("Missing value: ",df.isnull().sum())
print("Duplicate Value: ",df.duplicated().sum())
```

```python
#Remove Dublicate rows:
df = df.drop_duplicates()

print("After Remove duplicate: ", df.shape)
```

```python
df
```

```python
#Fill Missing amount using City:

df['amount'] = df.groupby('city')['amount'].transform(lambda x: x.fillna(x.mean()))
df
```

```python
#Verify Missing Value:
print("Missing amount after cleaning: ", df['amount'].isnull().sum())
```

3.Assignment 3 — Inner vs Left Merge

```python
#take ass 2 reference dataset in department column take :
depart_col = pd.DataFrame(df2['department'])
```

```python
depart_col  
```

```python
#Remove duplicate :
depart_col = depart_col.drop_duplicates()

depart_col
```

```python
#Left Merge:
left_merge = pd.merge(
    df2,
    depart_col,
    on="department",
    how="left"
)

print(left_merge)
print("Rows:", len(left_merge))
```

```python
#Inner Merge:
inner_merge = pd.merge(
    df2,
    df2['department'],
    on="department",
    how="inner"
)

print(inner_merge)
print("Rows:", len(left_merge))
```

Assignment 4 — Filtering with .isin()

```python
#Calculate department avg: 
dept_avg = df2.groupby("department")["marks"].mean()

print(dept_avg)
```

```python
#Filtering depart:
filtered = df2[
    df2["department"].isin(["Computer Science", "Mechanical"])
]

filtered
```

```python
#Comper with dipartment avg, also create department_avg col:
df2["department_avg"] = df2["department"].map(dept_avg)
```

```python
df2
```

```python
result = df2[
    df2["department"].isin(["Computer Science", "Mechanical"])
    & (df2["marks"] < df2["department_avg"])
]

print(result[["name", "department", "marks", "department_avg"]])
```
